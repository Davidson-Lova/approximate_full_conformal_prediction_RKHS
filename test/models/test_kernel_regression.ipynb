```python
import numpy as np
from time import time
from sklearn.metrics.pairwise import pairwise_kernels
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.kernel_ridge import KernelRidge

import sys
sys.path.append("../..")
```

```python
from src.data.load_data import load_data
from src.models.kernel_regression import KernelRegression
from src.models.kernel_empirical_risk import KernelEmpiricalRisk
```

Load data

```python
input_points, output_points = load_data("friedman1")
```

```python
train_input_points, test_input_points, train_output_points, test_output_points = (
    train_test_split(input_points, output_points, random_state=0)
)

scaler = StandardScaler()
scaler.fit(train_input_points)
train_input_points = scaler.transform(train_input_points)
test_input_points = scaler.transform(test_input_points)

# scaler_y = StandardScaler()
# scaler_y.fit(train_output_points)
# train_output_points = scaler_y.transform(train_output_points)
# test_output_points = scaler_y.transform(test_output_points)
```

Fit with lbfgs

```python
predictor = KernelRegression(solver="newton-cholesky",loss_name="quadratic", loss_params={"dmax":5.}, lam=0.5)
predictor.fit(train_input_points, train_output_points)
test_predicted_output_points = predictor.predict(test_input_points)
print("RMSE:", np.sqrt(np.mean((test_predicted_output_points - test_output_points)**2)))
```

```python
predictor_simple = KernelRidge(alpha=1.)
predictor_simple.fit(train_input_points, train_output_points)
test_predicted_output_points = predictor_simple.predict(test_input_points)
print("RMSE:", np.sqrt(np.mean((test_predicted_output_points - test_output_points)**2)))
```

Fit with newton

```python
predictor = KernelRegression(solver="lbfgs")
predictor.fit(train_input_points, train_output_points)
test_predicted_output_points = predictor.predict(test_input_points)
print("RMSE:", np.sqrt(np.mean((test_predicted_output_points - test_output_points)**2)))
```
